# Перевірка середовища: Gemini API у Google Colab
Ключ API зберігається в **Colab → Secrets** (`GEMINI_API_KEY` або `GOOGLE_API_KEY`) і не потрапляє в код.

```
!pip -q install -U google-genai
```

```
import os
from google import genai

def load_api_key():
    """Ключ береться з Colab Secrets або з файлу .env (локальний запуск)."""
    try:
        from google.colab import userdata
        for name in ("GEMINI_API_KEY", "GOOGLE_API_KEY"):
            try:
                return userdata.get(name)
            except Exception:
                pass
    except ImportError:
        from dotenv import load_dotenv
        load_dotenv()
    return os.getenv("GEMINI_API_KEY") or os.getenv("GOOGLE_API_KEY")

client = genai.Client(api_key=load_api_key())
print("Клієнт Gemini API успішно ініціалізовано!")
```

```
for m in client.models.list():
    actions = ",".join(m.supported_actions or [])
    if "gemini" in m.name or "embed" in m.name:
        print(f"{m.name:45} {actions}")
```
